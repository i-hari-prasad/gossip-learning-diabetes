# Evaluation Metrics
This notebook loads and prints the obtained metrics from the `results/` directory.

```python
import json
import pandas as pd

with open('../results/lglm_evaluation.json', 'r') as f:
    lglm = json.load(f)
    
with open('../results/centralized_lstm_evaluation.json', 'r') as f:
    lstm = json.load(f)

print('=== OVERALL METRICS ===')
print(f"LGLM MARD: {lglm['overall']['MARD']:.2f}%")
print(f"LGLM RMSE: {lglm['overall']['RMSE']:.3f}")
print(f"LGLM R2:   {lglm['overall']['R2']:.3f}")
print()
print('=== CLARKE ERROR GRID (LGLM) ===')
for k, v in lglm['clarke_zones'].items():
    print(f"{k}: {v:.2f}%")
```
